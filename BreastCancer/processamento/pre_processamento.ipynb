```python
import pandas as pd

df = pd.read_csv("wdbc.data", decimal="m", header=None)
display(df)
```

```python
df_tratada[2] = pd.to_numeric(df_tratada[2])
alvos = df_tratada.columns[2:]
df_tratada[alvos] = df_tratada[alvos].apply(pd.to_numeric)
df_tratada.info()
```
